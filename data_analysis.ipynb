```python
import numpy as np

# 1D array
a = np.array([1, 2, 3])
print(a)  # Output: [1 2 3]
print(a.ndim)  # 1 (1-dimensional)
print(a.shape)  # (3,) → 3 elements in 1D

# 2D array
b = np.array([[1, 2, 3], [4, 5, 6]])
print(b)  # Output: [[1 2 3]
          #          [4 5 6]]
print(b.ndim)  # 2 (2-dimensional)
print(b.shape)  # (2, 3) → 2 rows, 3 columns

# 3D array
c = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(c.shape)  # (2, 2, 2) → 2 blocks, 2 rows, 2 cols
print(c.ndim)  # 3 (3-dimensional)
```
